# Unseen Data Analysis

This notebook explores the unseen, unlabeled dataset located in `Data/unseen/un-labelled.csv`.


```python
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
import warnings
warnings.filterwarnings('ignore')

# Ensure plots directory exists
import os
os.chdir('/home/ashrafs/projects/Conspiracy-Proj')
Path('plots').mkdir(exist_ok=True)

# Plotting style
try:
    plt.style.use('seaborn-v0_8-darkgrid')
except:
    plt.style.use('seaborn-darkgrid')
sns.set_palette('husl')

print('Libraries imported. Working directory:', os.getcwd())
```

## 1. Load Unseen Dataset


```python
# Load unseen dataset
unseen_path = 'Data/unseen/un-labelled.csv'
df_unseen = pd.read_csv(unseen_path)

print('Rows:', len(df_unseen))
print('Columns:', df_unseen.columns.tolist()[:10], '...')
print('\nDataset sample:')
print(df_unseen.head())
```

## 2. Basic Dataset Statistics


```python
# Total number of messages and unique channels
if 'name' not in df_unseen.columns:
    raise ValueError("Expected column 'name' for channel information not found.")

total_messages = len(df_unseen)
unique_channels = df_unseen['name'].nunique()

print("Total number of messages:", total_messages)
print("Total unique channels:", unique_channels)
```

```python
# Total messages grouped by channel
messages_per_channel = df_unseen.groupby('name')['id'].count().sort_values(ascending=False)
print("Top 10 channels by message count:")
print(messages_per_channel.head(10))

# Save full distribution to CSV for reference
messages_per_channel.to_csv('Data/unseen/messages_per_channel.csv', header=['message_count'])
print("\nFull channel distribution saved to Data/unseen/messages_per_channel.csv")
```

```python
# Range of message dates
df_unseen['date_parsed'] = pd.to_datetime(df_unseen['date'], errors='coerce')
min_date = df_unseen['date_parsed'].min()
max_date = df_unseen['date_parsed'].max()

print("Date range for messages:")
print("  Earliest message:", min_date)
print("  Latest message:", max_date)
print(f"  Total days covered: {(max_date - min_date).days if pd.notna(min_date) and pd.notna(max_date) else 'N/A'}")

missing_dates = df_unseen['date_parsed'].isna().sum()
if missing_dates:
    print(f"\nNote: {missing_dates} messages have invalid or missing dates.")
```
